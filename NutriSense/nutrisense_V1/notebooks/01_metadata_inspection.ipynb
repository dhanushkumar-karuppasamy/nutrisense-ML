# 01. Metadata Inspection

This notebook inspects the variable names, labels, and category codes (value labels) from the Kids Recode (KR), Household Recode (HR), and Individual Recode (IR) datasets.

```python
import os
import pyreadstat
```

## 1. Verify State Code (`v024`) for Tamil Nadu

Before filtering, we verify the exact state code mapping from the DTA file metadata.

```python
raw_kr_path = os.path.join("..", "data", "raw", "IAKR7EFL.DTA")
_, meta_kr = pyreadstat.read_dta(raw_kr_path, metadataonly=True)

# Look up value labels for state variable v024
v024_labels = meta_kr.variable_value_labels.get('v024')
if v024_labels:
    print("Value labels for v024 (State):")
    for code, label in sorted(v024_labels.items()):
        if 'tamil' in label.lower():
            print(f"-> FOUND TAMIL NADU: Code {code} = {label}")
        else:
            print(f"   Code {code} = {label}")
else:
    print("No value labels found for v024.")
```

## 2. Inspect Wealth Index (`v190`) and Mother's Education (`v106`)

Let's also look up the category labels for the wealth index and education level to understand how they are coded.

```python
for var in ['v106', 'v190']:
    labels = meta_kr.variable_value_labels.get(var)
    if labels:
        print(f"\nValue labels for {var} ({meta_kr.column_names_to_labels.get(var)}):")
        for code, label in sorted(labels.items()):
            print(f"  {code}: {label}")
```
